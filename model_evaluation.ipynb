## Summary

This evaluation provides comprehensive performance metrics for your facial emotion detection model:

- **Accuracy**: Overall correctness of predictions
- **Precision**: When model predicts an emotion, how often is it correct?
- **Recall**: Of all actual instances of an emotion, how many does the model find?
- **F1 Score**: Harmonic mean of precision and recall
- **ROC AUC**: Measures the model's ability to distinguish between emotion classes
- **Confusion Matrix**: Shows which emotions the model confuses with each other

All visualizations have been saved as PNG files for reference.

```python
# Section 7: ROC AUC Analysis
print("="*70)
print("ROC AUC ANALYSIS")
print("="*70)

# Compute ROC AUC for multi-class
y_test_bin = label_binarize(y_test, classes=range(NUM_EMOTIONS))

try:
    roc_auc_weighted = roc_auc_score(y_test_bin, y_pred_proba, average='weighted', multi_class='ovr')
    print(f"\nROC AUC Score (weighted): {roc_auc_weighted:.4f}")
except Exception as e:
    print(f"\nROC AUC calculation skipped: {str(e)}")

# Plot ROC curves for each emotion class
fig, ax = plt.subplots(figsize=(12, 8))
colors = plt.cm.Set3(np.linspace(0, 1, NUM_EMOTIONS))

for i, emotion in enumerate(EMOTIONS):
    fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_pred_proba[:, i])
    roc_auc = auc(fpr, tpr)
    ax.plot(fpr, tpr, label=f'{emotion} (AUC = {roc_auc:.3f})', 
            color=colors[i], linewidth=2)

# Plot random classifier line
ax.plot([0, 1], [0, 1], 'k--', linewidth=2, label='Random Classifier (AUC = 0.5)')

ax.set_xlim([0.0, 1.0])
ax.set_ylim([0.0, 1.05])
ax.set_xlabel('False Positive Rate', fontsize=12)
ax.set_ylabel('True Positive Rate', fontsize=12)
ax.set_title('ROC Curves - Emotion Detection Model (One-vs-Rest)', fontsize=14, fontweight='bold')
ax.legend(loc="lower right", fontsize=10)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('roc_curves.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ ROC curves saved as 'roc_curves.png'")
```
